```python
# same inside notebook
import io
import soundfile as sf
from IPython.display import Audio, display

i = 0

# Get one audio sample
audio = df["cleaned_audio"].iloc[i]

# Decode audio
data, sr = sf.read(io.BytesIO(audio["bytes"]))

# Audio statistics
duration = len(data) / sr

print("Sample index:", i)
print("Sample rate :", sr)
print("Shape       :", data.shape)
print("Channels    :", 1 if data.ndim == 1 else data.shape[1])
print("Duration    :", duration, "sec")
print("Min value   :", data.min())
print("Max value   :", data.max())
print("Bytes       :", len(audio["bytes"]))
print("Path        :", audio["path"])
print("Bytes type  :", type(audio["bytes"]))
```

so resampling is correct

```python
# lets listen one sample

import subprocess
from pathlib import Path

i = 0

# Get audio + text
text = df.iloc[i]["raw_text"]
audio_bytes = df.iloc[i]["cleaned_audio"]["bytes"]

# Save audio
path = Path(f"sample_{i}.ogg")
path.write_bytes(audio_bytes)

print("Text:", text)
print("Opening:", path.absolute())

# Open with the system's default media player
subprocess.Popen(["xdg-open", str(path.absolute())])
```
