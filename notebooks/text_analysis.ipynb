```python
import re
import io
import tempfile
import subprocess
import soundfile as sf

patterns = {
    "uppercase": r"\b[A-Z]{2,}\b",
    "dotted": r"\b(?:[A-Za-z]\.){2,}",
}

count = 0

for _, row in df.iterrows():

    text = row["text"]

    if (
        re.search(patterns["uppercase"], text)
        or re.search(patterns["dotted"], text)
    ):
        print(text)

        audio = row["audio"]
        data, sr = sf.read(io.BytesIO(audio["bytes"]))

        with tempfile.NamedTemporaryFile(suffix=".wav") as f:
            sf.write(f.name, data, sr)

            subprocess.run([
                "ffplay",
                "-nodisp",
                "-autoexit",
                "-loglevel", "quiet",
                f.name
            ])

        print("-" * 80)

        count += 1
        if count == 5:
            break
```

```python
import re
import io
import tempfile
import subprocess
import soundfile as sf

pattern = re.compile(
    r"\b(?:kg|km|cm|mm|mg|ml|Dr|Mr|Mrs|Ms|etc|AQSH|BMT|OAV|MDH)\b"
    r"|%|\b(?:[A-Za-zА-Яа-яЎўҒғҚқҲҳЁё]\.){2,}",
    re.IGNORECASE
)

matches = []

for i, row in df.iterrows():
    if pattern.search(row["text"]):
        matches.append((i, row))

# First 2 + last 2
selected = matches[:2] + matches[-2:]

for i, row in selected:
    text = row["text"]
    print(f"\n[{i}] {text}")

    # Decode audio
    audio = row["audio"]
    data, sr = sf.read(io.BytesIO(audio["bytes"]))

    # Save temporarily as WAV
    with tempfile.NamedTemporaryFile(suffix=".wav") as f:
        sf.write(f.name, data, sr)

        # Play through system speakers/headphones
        subprocess.run(
            ["ffplay", "-nodisp", "-autoexit", "-loglevel", "quiet", f.name]
        )
```

```python
#and
# 
import re
import unicodedata


def normalize_text(text: str) -> str:
    # 1. Unicode normalization
    text = unicodedata.normalize("NFC", text)

    # 2. Remove leading/trailing whitespace
    text = text.strip()

    # 3. Normalize repeated whitespace
    text = re.sub(r"\s+", " ", text)

    # 4. TODO:
    #    numbers
    #    abbreviations
    #    punctuation
    #    script conversion if required

    return text 
```
